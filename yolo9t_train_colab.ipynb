```python
# 1. Mount Google Drive
from google.colab import drive
import yaml
import shutil
import os
drive.mount('/content/drive')
```

```python
# 2. Unzip the dataset rapidly to the local Colab disk
print("Unzipping dataset to local Colab disk...")
!unzip -q "/content/drive/MyDrive/raya_dataset/yolo_dataset.zip" -d /content/yolo_dataset
```

```python
# 3. Fix data.yaml — the actual data is nested at /content/yolo_dataset/yolo_dataset/
data_yaml_path = '/content/yolo_dataset/yolo_dataset/data.yaml'
with open(data_yaml_path, 'r') as f:
    data = yaml.safe_load(f)
data['path'] = '/content/yolo_dataset/yolo_dataset'
data['train'] = 'images/train'
data['val'] = 'images/val'
with open(data_yaml_path, 'w') as f:
    yaml.dump(data, f, sort_keys=False)
print("data.yaml fixed! Contents:")
print(open(data_yaml_path).read())
```

```python
# 4. Verify the folders actually exist before training
assert os.path.isdir('/content/yolo_dataset/yolo_dataset/images/train'), "train folder missing!"
assert os.path.isdir('/content/yolo_dataset/yolo_dataset/images/val'), "val folder missing!"
print("✅ Both train and val folders confirmed.")
```

```python
# 5. Install Ultralytics
%pip install -q ultralytics
```

```python
# 6. Train
from ultralytics import YOLO
model = YOLO('yolov9t.pt')
results = model.train(
    data=data_yaml_path,
    epochs=100,
    imgsz=416,
    batch=64,
    device=0,
    patience=20,
    project='/content/drive/MyDrive/ANPR_Training',
    name='yolov9t_15k'
)
```
